### Project Highlights:
1. **Problem Formulation:** Supervised Regression predicting continuous Opening Weekend Gross ($ Millions).
2. **Top 10 Research-Backed Factors:** Opening Theater Capacity, Production Budget, Market Competition / Cannibalization, Franchise Status, Genre, Studio Clout, Summer/Holiday Seasonality, Runtime, and Concept Scope.
3. **Zero Lookahead Leakage:** Strictly pre-release predictors (no post-release audience reviews or ratings).
4. **Methodology:** Mean Baseline + Ridge Regression + Random Forest + Tuned Gradient Boosting with 5-Fold Cross Validation.
5. **State-of-the-Art Results:** Machine learning models capture over **78.5% of the variance ($R^2 = 0.785$)** with a **Median Error of just $4.17M**.
